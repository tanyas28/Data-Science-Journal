```python
import warnings
warnings.filterwarnings("ignore")
#step 2 : generating sentnece embeding
from sentence_transformers import SentenceTransformer
sentences = [s1,s2,s3,s4,s5,s6,s7,s8,s9,s10]

embeding = SentenceTransformer('all-MiniLM-L6-v2')
sentence_embeddings = embeding.encode(sentences)
print(sentence_embeddings.shape)
```

```python
#now cosine simalarity
import numpy as np
from numpy.linalg import norm
def cosine_smil(a,b):
  dot_prod = np.dot(a,b)
  norm_a = norm(a)
  norm_b = norm(b)
  return dot_prod / (norm_a * norm_b)
```

```python
# we need embeding simalrity score between query and our db
query ="What is Harry Potter?"
query_embeding = embeding.encode(query)
print(query_embeding.shape)
```

```python
score_array = []
for i in range (0,10,1):
  score = cosine_smil(sentence_embeddings[i],query_embeding)
  score_array.append(score)
print(score_array)


```

```python
top_3_indices = np.argsort(score_array)[-3:]
print(top_3_indices)
```

```python
for i in top_3_indices:
  print(sentences[i],score_array[i])
```
