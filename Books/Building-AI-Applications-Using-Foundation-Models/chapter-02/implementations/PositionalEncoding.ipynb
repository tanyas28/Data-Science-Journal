```python
import numpy as np
pos =11 # to correspond to number of words used in sentence in the implementation of attention function
d=4 # to correspond to the embedding dimension used in the implementation of attention function

PE=np.zeros((pos,d),dtype=np.float32)

print(PE)
print(PE.shape)
```

```python
for i in range(pos):
    for j in range(d):
        if j%2==0:
            PE[i,j]=np.sin(i/(10000**(j/d)))
        else:
            PE[i,j]=np.cos(i/(10000**((j-1)/d)))

print(PE)
```

That is it this is then added to embedding matrix and then rest of the attention step follow.
