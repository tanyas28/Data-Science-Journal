```python
from transformers import AutoModelForCausalLM, AutoTokenizer, AutoConfig
import torch

# Explicitly load the configuration first to ensure all parameters, including rope_scaling,
# are correctly parsed before passing to the model.
config = AutoConfig.from_pretrained("microsoft/Phi-3-mini-4k-instruct", trust_remote_code=True)

# Fix for KeyError: 'type' when rope_scaling is null in config.json but loaded as an invalid dict.
# Ensure config.rope_scaling is None if it's a dictionary that doesn't contain the 'type' key.
# This handles cases where 'null' is interpreted as an empty dict or an incomplete one.
if isinstance(config.rope_scaling, dict) and "type" not in config.rope_scaling:
    config.rope_scaling = None

model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",
    config=config, # Pass the pre-loaded configuration
    device_map = "cuda",
    torch_dtype="auto",
    trust_remote_code=True,
)

tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")
```

```python
prompt = "write an email to manager requesting for work from home."

#tokenizing prompt
input_ids = tokenizer(prompt, return_tensors="pt").input_ids.to("cuda")

#generate text
generation_output = model.generate(
    input_ids = input_ids,
    max_new_tokens = 25,
    use_cache=False
)

print(tokenizer.decode(generation_output[0]))
```

```python
print(input_ids) #this went into the model. tokenizer processed input and returned info that model needs.
```

these id values are actually token ids , they point to token id table , that has all the tokens that this tokenizer know.

```python
#to see the words we can decode

for id in input_ids[0]:
  print(tokenizer.decode(id))
```

```python
#lets see ouput token ids
print(generation_output)
```

it has both input and output token

```python
for out in generation_output:
  print(tokenizer.decode(out))
```

creating better token embeddings

```python
from transformers import AutoModel, AutoTokenizer

#load tokenizer
tokenizer=AutoTokenizer.from_pretrained("microsoft/deberta-base")

#load model
model = AutoModel.from_pretrained("microsoft/deberta-v3-xsmall")

#tokenize
tokens = tokenizer("Computer science is fun", return_tensors="pt")

#process token
output = model(**tokens)[0]
```

```python
output.shape
```

```python
for token in tokens['input_ids'][0]:
  print(tokenizer.decode(token))
```

```python
print(output)
```

now, in the above example we see each row a vector embedding for each token, but what we use now mostly actually represent sentences , paragraphs as single vector. lets see:

```python
from sentence_transformers import SentenceTransformer

#load model
model = SentenceTransformer("sentence-transformers/all-mpnet-base-v2")

#encode sentences
vector = model.encode("Ai is the most intresting domain in computer science")
```

```python
vector.shape # we see a single row for entore sentence with dimension 768.
```
